### House price dataset

```python
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt
%matplotlib inline
```

```python
housing = pd.read_csv("../Data/HousingDataset.csv")
```

```python
housing.head()
```

```python
housing.info()
```

```python
housing.describe()
```

```python
print(housing.isna().sum())
```

```python
# Price based on overall quality
print(housing.groupby('OverallQual')['SalePrice'].mean())
```

```python
sns.barplot(data=housing, x='OverallQual', y='SalePrice')
plt.xlabel('Overall Quality')
plt.ylabel('Sale Price')
plt.title('Sale Price Based on Overall Quality')
plt.savefig("../Results/price_by_quality.png")
plt.show()
```

```python
# Price based on neighborhood
print(housing.groupby('Neighborhood')['SalePrice'].mean())
```

```python
sns.barplot(data=housing, x='Neighborhood', y='SalePrice')
plt.xticks(rotation=90)
plt.xlabel('Neighborhood')
plt.ylabel('Sale Price')
plt.title('Sale Price Based on Neighborhood')
plt.savefig("../Results/price_by_neighborhood.png", bbox_inches='tight')
plt.show()
```

```python
# Price based on year-built group
housing['YearGroup'] = pd.cut(housing['YearBuilt'], bins=[1870, 1920, 1950, 1980, 2000, 2010],
                              labels=['Before 1920', '1920-1950', '1950-1980', '1980-2000', '2000+'])
```

```python
sns.barplot(data=housing, x='YearGroup', y='SalePrice')
plt.xlabel('Year Built')
plt.ylabel('Sale Price')
plt.title('Sale Price Based on Year Built')
plt.savefig("../Results/price_by_year_built.png")
plt.show()
```
